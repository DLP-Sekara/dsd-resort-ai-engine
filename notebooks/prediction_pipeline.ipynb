```python
import joblib
import pandas as pd
import numpy as np
```

```python

# 1. Load the saved model and column structure
model = joblib.load('../static/model/best_demand_model.pkl')
model_columns = joblib.load('../static/model/model_columns.pkl')


# 2. Simulate a prediction request for a specific future date (e.g., Tomorrow)
# Let's say we want to predict for: DayOfWeek=5 (Saturday), IsWeekend=1, IsHoliday=0, Temperature=30, Weather='Rainy'
input_data = {
    'DayOfWeek': 2,
    'IsWeekend': 0,
    'IsHoliday': 1,
    'Temperature': 27,
    'Weather': 'Rainy' # We will one-hot encode this just like training data
}

def predict_guests(data):
    # Convert input dict to DataFrame
    df_input = pd.DataFrame([data])
    
    # One-hot encode the 'Weather' column
    df_input = pd.get_dummies(df_input, columns=['Weather'], dtype=int)
    
    # Align columns with the training columns (fill missing weather columns with 0)
    for col in model_columns:
        if col not in df_input.columns:
            df_input[col] = 0
            
    # Reorder columns to match the exact order the model expects
    df_input = df_input[model_columns]
    
    # Make prediction using the loaded model
    prediction = model.predict(df_input)
    
    # Return rounded integer guest count
    return max(10, int(round(prediction[0])))

# Test the prediction function
predicted_count = predict_guests(input_data)
print(f" Predicted Guest Count for the given conditions: ~{predicted_count} guests")
```
